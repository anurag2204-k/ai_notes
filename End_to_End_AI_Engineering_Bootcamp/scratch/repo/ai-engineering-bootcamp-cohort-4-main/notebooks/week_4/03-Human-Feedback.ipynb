### Import Dependencies

```python
from langsmith import Client
```

### Attach Feedback to a specific trace in LangSmith

019d794b-554b-71e1-a542-d58c39517478

```python
client = Client()
```

```python
RUN_ID = "019d794b-554b-71e1-a542-d58c39517478"
```

```python
client.create_feedback(
    run_id=RUN_ID,
    key="feedback",
    score=0,
    feedback_source_type="api"
)
```

```python
client.create_feedback(
    run_id=RUN_ID,
    key="comment",
    value="This is a long form free text feedback left by Aurimas",
    feedback_source_type="api"
)
```

```python
client.create_feedback(
    run_id=RUN_ID,
    key="comments_with_score",
    value="This is a long form free text feedback left by Aurimas",
    score=1,
    feedback_source_type="api"
)
```
